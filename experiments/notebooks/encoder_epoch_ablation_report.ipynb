# Encoder-epoch ablation report

Grouped BO and external-reconstruction results for the retained encoder checkpoints. Values are mean ± 95% CI.

```python
import sys
import pandas as pd

sys.path.insert(0, "../..")

from experiments.plotting.encoder_epoch_ablation import DATA_PATH, epoch_summary

data = pd.read_csv(DATA_PATH)
data = data[data["checkpoint_exists"].astype(bool)].copy()
summary = epoch_summary(data)

def compact_report(problem, metrics):
    source = summary[summary["problem"] == problem].copy()
    report = source[["epoch", "n"]].rename(columns={"epoch": "Epoch", "n": "N"})
    for label, metric in metrics:
        report[label] = [
            f"{mean:.4f} ± {ci95:.4f}"
            for mean, ci95 in zip(source[f"{metric}_mean"], source[f"{metric}_ci95"])
        ]
    return report
```

## Engine

```python
engine_report = compact_report(
    "engine",
    [
        ("Gap ratio", "best_gap_ratio"),
        ("Gap regret", "best_gap_regret"),
        ("Adjacency F1", "bo_cutoff_adj_f1"),
        ("Feature MAE", "feature_mae"),
    ],
)
print(engine_report.to_string(index=False))
```

## MDGNC

```python
mdgnc_report = compact_report(
    "mdgnc",
    [
        ("ΔHV ratio", "delta_hv_ratio"),
        ("ΔHV regret", "delta_hv_regret"),
        ("Adjacency F1", "bo_cutoff_adj_f1"),
        ("Feature MAE", "feature_mae"),
    ],
)
print(mdgnc_report.to_string(index=False))
```
